```python
import cv2
import hashlib
import numpy as np
from Crypto.Cipher import AES
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes

#ascii conversion
d = {chr(i): i for i in range(256)}
c = {i: chr(i) for i in range(256)}


def derive_key(userkey):
    return hashlib.sha256(userkey.encode()).digest()[:16]  # 128-bit AES key

#Encryption
def encrypt_message(msg, userkey):
    key = derive_key(userkey)
    cipher = AES.new(key, AES.MODE_CBC)
    ct = cipher.encrypt(pad(msg.encode(), AES.block_size))
    return cipher.iv + ct  # prepend IV to ciphertext

#Decryption
def decrypt_message(cipher_bytes, userkey):
    key = derive_key(userkey)
    iv = cipher_bytes[:16]
    ct = cipher_bytes[16:]
    cipher = AES.new(key, AES.MODE_CBC, iv)
    return unpad(cipher.decrypt(ct), AES.block_size).decode()

#Load Image
x = cv2.imread('rani.jpg')
if x is None:
    raise FileNotFoundError("Image 'rani.jpg' not found in the current directory.")

#Input
key = "12"
text = "ranijhari"


encrypted_bytes = encrypt_message(text, key)
l = len(encrypted_bytes)
print(f"Encrypted byte length: {l}")

#Flatten image
x_enc = x.copy()
flat = x_enc.flatten()

if l > len(flat):
    raise ValueError("Message too large to embed in image.")

#combinning msg and image
for i in range(l):
    flat[i] = encrypted_bytes[i] ^ d[key[i % len(key)]]

# Reshape and save
x_enc = flat.reshape(x_enc.shape)
cv2.imwrite("encrypting.jpg", x_enc)
print(" Encrypted image saved as 'encrypting.jpg'")
import sys
import subprocess
import os

def open_image(filename):
    if sys.platform.startswith('darwin'):
        subprocess.call(('open', filename))
    elif os.name == 'nt':
        os.startfile(filename)
    elif os.name == 'posix':
        subprocess.call(('xdg-open', filename))



#Decryption
key1 = input(" Re-enter key for decryption: ")
if key1 != key:
    print(" Wrong key entered. Access denied.")
else:
    encrypted_back = bytearray()
    for i in range(l):
        byte = x_enc.flatten()[i] ^ d[key1[i % len(key1)]]
        encrypted_back.append(byte)

    try:
        decrypted = decrypt_message(bytes(encrypted_back), key1)
        print(" Decrypted message:", decrypted)
    except Exception as e:
        print(" Decryption failed:", str(e))
```
